##### 직접 만든 가위바위보 탐지 모델을 사용해보기

```python
from ultralytics import YOLO  # ultralytics란 큰 라이브러리에서 YOLO란 작은 라이브러리 불러오기 
import cv2 # opencv라는 라이브러리 불러오기
import os # os라는 라이브러리 불러오기

model = YOLO('custom_model.pt') # 객체 탐지 모델 정의하기(pt파일은 모델이 저장되는 형식)

# 탐지할 파일 정의하기 영상, 사진
#file_path = 'tests/test1.mp4'  
file_path = 'tests/1.jpg'

# 사진 혹은 영상의 프레임 1장씩 모델 예측하는 함수
def process_frame(frame):
    results = model(frame)
    annotated_frame = results[0].plot()

    new_w = 756
    new_h = 1008
    resized_frame = cv2.resize(annotated_frame, (new_w, new_h))
    cv2.imshow('YOLOv8 Object Detection', resized_frame)

# 탐지할 파일의 형식에 대한 정보 추출하기 mp4인지 jpg, png인지 구별
file_ext = os.path.splitext(file_path)[-1].lower()

# 조건문 사진 형식을 입력했을경우
if file_ext in ['.jpg', '.jpeg', '.png', '.bmp']: 
    image = cv2.imread(file_path)
    if image is None:
        print("Error: 이미지를 열 수 없습니다.")
        exit()
    process_frame(image)
    cv2.waitKey(0) 
    cv2.destroyAllWindows()

# 조건문 영상 형식을 입력했을 경우
elif file_ext in ['.mp4', '.avi', '.mov', '.mkv']: 
    cap = cv2.VideoCapture(file_path)
    if not cap.isOpened():
        print("Error: 동영상을 열 수 없습니다.")
        exit()

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        process_frame(frame)
        if cv2.waitKey(1) & 0xFF == ord('q'): 
            break

    cap.release()
    cv2.destroyAllWindows()

# 사진, 영상 그 어떤 형식도 아닌 경우
else:
    print("Error: 지원하지 않는 파일 형식입니다.")
```
